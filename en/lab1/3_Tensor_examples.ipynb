```python
import torch

# order 0 tensors (scalars)
number = torch.tensor(5, dtype=torch.int16)
print("`number` is a {}-d Tensor".format(number.ndim))

# order 1 tensors
primes = torch.tensor([2, 3, 5, 7, 11], dtype=torch.int16)
print("`primes` is a {}-d Tensor".format(primes.ndim))

# order 2 (matrices)
matrix = torch.tensor([[3.1, 8.2, 5.9], [4.3, -2.7, 6.5]], dtype=torch.float64)
print("`matrix` is a {}-d Tensor".format(matrix.ndim))
```

```python
tensor4 = torch.zeros ([3, 2, 4, 5])
print("Dtype of each element:", tensor4.dtype)
print("Number of axes:", tensor4.ndim)
print("Shape:", tensor4.shape)
print("Elements along axis 0:", tensor4.shape[0])
print("Elements along the last axis (3):",tensor4.shape[-1])
print("Total number of elements (3*2*4*5): ", tensor4.numel())
```
