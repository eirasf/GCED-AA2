```python
import numpy as np
import torch

np_array = np.array([1, 2, 3], dtype=np.float32)

# Option 1: shares memory
t1 = torch.from_numpy(np_array)
# Option 2: makes a copy
t2 = torch.tensor(np_array)

# Mutate the NumPy array
np_array[0] = 99
print(t1)  # tensor([99.,  2.,  3.])  <-- gets updated
print(t2)  # tensor([1., 2., 3.])     <-- does not change
```
