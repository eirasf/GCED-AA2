# Regularization

Once overfitting has been diagnosed, it is time to try different techniques that try to reduce the variance, without increasing the bias too much, and thus the model generalizes better. The regularization techniques that we are going to see in this lab are:
1. *Early stopping*. Stops the training of the network when the error increases.
1. Parameter norm penalty (both L1 and L2 norm).
1. *Dropout*. Widely used in deep learning, it "deactivates" some neurons to avoid overfitting.
